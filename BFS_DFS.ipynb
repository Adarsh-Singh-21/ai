# PR2: Uninformed Search (BFS and DFS on a Maze)

**Aim:** Solve a maze with Breadth First Search and Depth First Search and compare them.

| | BFS | DFS |
|---|---|---|
| Data structure | Queue (first in, first out) | Stack (last in, first out) |
| Shortest path? | Yes | Not always |
| Memory | More | Less |

## Maze

`S` = start, `E` = end, `#` = wall, `.` = free path.

## Helper: get valid neighbours (up, down, left, right)

```python
def get_neighbours(pos):
    r, c = pos
    result = []
    for dr, dc in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
        nr = r + dr
        nc = c + dc
        if 0 <= nr < len(maze) and 0 <= nc < len(maze[0]):
            if maze[nr][nc] != "#":
                result.append((nr, nc))
    return result
```

## BFS

Uses a **queue**. `pop(0)` takes from the front, so it explores level by level.

```python
def bfs(start, end):
    queue = [[start]]          # queue of paths
    visited = [start]
    explored = 0

    while len(queue) > 0:
        path = queue.pop(0)    # take from FRONT
        node = path[-1]
        explored = explored + 1

        if node == end:
            return path, explored

        for n in get_neighbours(node):
            if n not in visited:
                visited.append(n)
                queue.append(path + [n])

    return None, explored
```

## DFS

Uses a **stack**. `pop()` takes from the back, so it goes deep first.

(Only one line differs from BFS: `queue.pop(0)` becomes `stack.pop()`.)

```python
def dfs(start, end):
    stack = [[start]]          # stack of paths
    visited = [start]
    explored = 0

    while len(stack) > 0:
        path = stack.pop()     # take from BACK
        node = path[-1]
        explored = explored + 1

        if node == end:
            return path, explored

        for n in get_neighbours(node):
            if n not in visited:
                visited.append(n)
                stack.append(path + [n])

    return None, explored
```

## Compare: path length, nodes explored, time

```python
t1 = time.perf_counter()
bfs_path, bfs_nodes = bfs(start, end)
bfs_time = (time.perf_counter() - t1) * 1000

t2 = time.perf_counter()
dfs_path, dfs_nodes = dfs(start, end)
dfs_time = (time.perf_counter() - t2) * 1000

print("Algorithm | Path length | Nodes explored | Time (ms)")
print("BFS       |", len(bfs_path), "         |", bfs_nodes, "           |", round(bfs_time, 4))
print("DFS       |", len(dfs_path), "         |", dfs_nodes, "           |", round(dfs_time, 4))
```

## Show the path on the maze (`*` marks the path)

```python
def show_path(path):
    grid = [list(row) for row in maze]
    for (r, c) in path:
        if grid[r][c] == ".":
            grid[r][c] = "*"
    for row in grid:
        print("".join(row))

print("BFS path:")
show_path(bfs_path)
print()
print("DFS path:")
show_path(dfs_path)
```

```python
rows = int(input("Enter number of rows: "))
cols = int(input("Enter number of columns: "))

maze = []

for i in range(rows):
    while True:
        row = list(input("Enter row (. = free, # = wall): ").replace(" ", ""))

        if len(row) == cols and all(x in (".", "#") for x in row):
            maze.append(row)
            break
        else:
            print(f"Invalid input! Enter exactly {cols} characters using . and #.")

print("Maze:")
for row in maze:
    print("".join(row))

start = tuple(map(int, input("Enter start (row column): ").split()))
end = tuple(map(int, input("Enter end (row column): ").split()))
```

## Viva points

- **BFS** is complete and optimal (when every step costs the same). It finds the shortest path.
- **DFS** is complete only if we avoid revisiting nodes. It is not optimal, so its path can be longer.
- BFS time and space: O(b^d). DFS space: O(b*m), which is smaller.
- b = branching factor, d = depth of the goal, m = maximum depth.
