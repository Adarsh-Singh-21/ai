```python
# Simple Reflex Agent

def reflex_response(text):
    # Actuator
    if "hi" in text or "hello" in text:
        return "Hi! Ask me about Web Dev, Gen-AI, or ML courses."

    elif "courses" in text:
        return "We offer 3 courses - Web Dev, Gen-AI, and Machine Learning."

    elif "web dev" in text or "webdev" in text:
        return "Web Development - 12 weeks, Rs.8999. Topics: HTML, CSS, JS, React."

    elif "gen ai" in text or "genai" in text or "gen-ai" in text or "generative ai" in text:
        return "Generative AI - 8 weeks, Rs.12999. Topics: Prompt Engineering, RAG, LLMs."

    elif "machine learning" in text or text == "ml":
        return "Machine Learning - 10 weeks, Rs.10999. Topics: Python, Regression, Neural Networks."

    elif "price" in text or "cost" in text:
        return "Please tell me which course - Web Dev, Gen-AI, or ML."

    elif "duration" in text or "how long" in text:
        return "Please tell me which course - Web Dev, Gen-AI, or ML."

    elif "curriculum" in text or "learn" in text:
        return "Please tell me which course - Web Dev, Gen-AI, or ML."

    else:
        return "Sorry, I didn't understand. Try asking about Web Dev, Gen-AI, or ML."


```

```python
# Model Based Agent
memory = {"current_course": None, "user_name": None}

def model_response(text, memory):
    # Actuator
    if "my name is" in text:
        name = text.split("my name is")[-1].strip()
        memory["user_name"] = name
        return f"Nice to meet you, {name}!"

    elif "hi" in text or "hello" in text:
        if memory["user_name"]:
            return f"Hi {memory['user_name']}! Ask me about Web Dev, Gen-AI, or ML courses."
        return "Hi! Ask me about Web Dev, Gen-AI, or ML courses."

    elif "courses" in text:
        return "We offer 3 courses - Web Dev, Gen-AI, and Machine Learning."

    elif "web dev" in text or "webdev" in text:
        memory["current_course"] = "webdev"
        return "Web Development - 12 weeks, Rs.8999. Topics: HTML, CSS, JS, React."

    elif "gen ai" in text or "genai" in text or "gen-ai" in text or "generative ai" in text:
        memory["current_course"] = "genai"
        return "Generative AI - 8 weeks, Rs.12999. Topics: Prompt Engineering, RAG, LLMs."

    elif "machine learning" in text or text == "ml":
        memory["current_course"] = "ml"
        return "Machine Learning - 10 weeks, Rs.10999. Topics: Python, Regression, Neural Networks."

    elif "price" in text or "cost" in text:
        if memory["current_course"] == "webdev":
            return "Web Development costs Rs.8999."
        elif memory["current_course"] == "genai":
            return "Generative AI costs Rs.12999."
        elif memory["current_course"] == "ml":
            return "Machine Learning costs Rs.10999."
        else:
            return "Please tell me which course - Web Dev, Gen-AI, or ML."

    elif "duration" in text or "how long" in text:
        if memory["current_course"] == "webdev":
            return "Web Development takes 12 weeks."
        elif memory["current_course"] == "genai":
            return "Generative AI takes 8 weeks."
        elif memory["current_course"] == "ml":
            return "Machine Learning takes 10 weeks."
        else:
            return "Please tell me which course - Web Dev, Gen-AI, or ML."

    elif "curriculum" in text or "learn" in text:
        if memory["current_course"] == "webdev":
            return "Web Development covers HTML, CSS, JS, React."
        elif memory["current_course"] == "genai":
            return "Generative AI covers Prompt Engineering, RAG, LLMs."
        elif memory["current_course"] == "ml":
            return "Machine Learning covers Python, Regression, Neural Networks."
        else:
            return "Please tell me which course - Web Dev, Gen-AI, or ML."

    else:
        return "Sorry, I didn't understand. Try asking about Web Dev, Gen-AI, or ML."
```

```python
# Environment

mode = "reflex"
print("Bot: Welcome! I'm your course assistant.")
print("Bot: Type 'toggle' to switch between Reflex and Model-Based mode. Type 'exit' to leave.")

while True:
    # Sensor
    text = input(f"You : ").lower()

    if text == "exit" or text == "quit" or text == "bye":
        print("Bot: Goodbye!")
        break

    elif text == "toggle":
      if mode == "reflex":
        mode = "model"
      else:
          mode = "reflex"
      print(f"Bot: Switched to {mode} mode.")

    elif mode == "reflex":
        print("Bot:", reflex_response(text))

    else:
        print("Bot:", model_response(text, memory))
```
